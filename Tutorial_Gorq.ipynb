```python
from langchain.prompts import PromptTemplate
from langchain.llms import OpenAI
from langchain.chains import LLMChain
from langchain.output_parsers import StructuredOutputParser, ResponseSchema

# Step 1: Define a Prompt for UML Information Extraction
response_schemas = [
    ResponseSchema(name="classes", description="List of classes with attributes and methods."),
    ResponseSchema(name="relationships", description="List of relationships between classes with their types.")
]

output_parser = StructuredOutputParser.from_response_schemas(response_schemas)

prompt_template = PromptTemplate(
    input_variables=["text"],
    template=(
        "Extract UML components from the following text:\n{text}\n\n"
        "Return the result as:\n"
        "1. Classes: A list of classes, each with their attributes and methods.\n"
        "2. Relationships: A list of relationships between classes with their type.\n"
        "{format_instructions}"
    ),
    partial_variables={"format_instructions": output_parser.get_format_instructions()}
)

# Step 2: Initialize LLM and Chain
llm = OpenAI(model="llama-3.1-70b-versatile", temperature=0)  # Replace with your LLM
chain = LLMChain(llm=llm, prompt=prompt_template)

# Step 3: Process Input Text
input_text = (
    "Create a UML diagram for a Library Management System. "
    "It should include classes: Book (title, author, ISBN), Author (name, bio), "
    "and Member (name, membershipID). Relationships are: "
    "Book is written by Author, and Member borrows Book."
)

# Extract UML components
response = chain.run({"text": input_text})
parsed_response = output_parser.parse(response)

# Step 4: Display Extracted Information
print("Classes:", parsed_response["classes"])
print("Relationships:", parsed_response["relationships"])

# Step 5: (Optional) Use Graphviz to Visualize UML
from graphviz import Digraph

dot = Digraph(comment="UML Diagram")
for cls in parsed_response["classes"]:
    dot.node(cls["name"], f"{cls['name']}\nAttributes: {', '.join(cls['attributes'])}\nMethods: {', '.join(cls['methods'])}")

for rel in parsed_response["relationships"]:
    dot.edge(rel["source"], rel["target"], rel["type"])

# Save UML Diagram
dot.render("uml_diagram", format="png", cleanup=True)
```
